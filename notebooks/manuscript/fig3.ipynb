```python
def plot_loss_curve(ax, loss_data, label, color, show_marker_labels=True, show_minimum=True):
    # Columns are epoch and loss; discard epoch 0 and any non-finite rows.
    epochs = np.asarray(loss_data[:, 0])
    losses = np.asarray(loss_data[:, 1])
    valid = (epochs > 0) & np.isfinite(epochs) & np.isfinite(losses)
    epochs = epochs[valid]
    losses = losses[valid]

    ax.plot(epochs, losses, color=color, linewidth=1.2, label=label)

    # Mark the recorded point nearest epoch 70.
    epoch70_idx = np.argmin(np.abs(epochs - 70))
    ax.scatter(
        epochs[epoch70_idx], losses[epoch70_idx],
        marker='*', s=70, color=color, edgecolor='black', linewidth=0.6,
        zorder=4, label='Epoch 70' if show_marker_labels else None,
    )

    # Mark the minimum recorded loss with a different symbol.
    if show_minimum:
        minimum_idx = np.argmin(losses)
        ax.scatter(
            epochs[minimum_idx], losses[minimum_idx],
            marker='D', s=24, facecolor='white', edgecolor=color, linewidth=1.2,
            zorder=5, label='Minimum loss',
        )


f, ax = plt.subplots(1, 5, figsize=(7, 2), dpi=300, sharey=True)
f.subplots_adjust(left=0.07, right=0.995, bottom=0.22, top=0.86, wspace=0.10)

plot_loss_curve(ax[0], lsqml_loss, 'Default LSQML', 'tab:blue')
ax[0].set_title('Default LSQML', fontsize=8, pad=3)

plot_loss_curve(ax[1], epie_loss, 'ePIE', 'tab:red', show_minimum=False)
plot_loss_curve(ax[1], long_epie_loss, 'Long ePIE', 'tab:red', show_marker_labels=False)
ax[1].set_title('ePIE', fontsize=8, pad=3)

plot_loss_curve(ax[2], object_loss, 'Object-only LSQML', 'tab:green')
ax[2].set_title('Object-only LSQML', fontsize=8, pad=3)

plot_loss_curve(ax[3], opr_loss, 'OPR', 'tab:orange')
ax[3].set_title('OPR', fontsize=8, pad=3)

plot_loss_curve(ax[4], tv_loss, 'Preprocessing + TV', 'tab:purple')
ax[4].set_title('Preprocessing + TV', fontsize=8, pad=3)


for axis in ax:
    axis.set_xlabel('Epoch', fontsize=7, labelpad=2)
    axis.tick_params(axis='both', which='both', direction='in', labelsize=6, length=2.5, width=0.6, pad=2)
    axis.yaxis.get_offset_text().set_fontsize(6)
    axis.grid(alpha=0.25, linewidth=0.4)
    axis.legend(fontsize=5.5, frameon=False, handlelength=1.2, handletextpad=0.4, labelspacing=0.25, borderaxespad=0.3)
    for spine in axis.spines.values():
        spine.set_linewidth(0.6)
ax[0].set_ylabel('Loss', fontsize=7, labelpad=2)

#f.savefig("losses.svg", transparent=True, bbox_inches="tight", dpi=300)
```
